## 1. Calculate the difference

Predict the output before running the cell. The first two lines give names to the quantities. The third calculates and names their difference. `print` displays that result. Change ten to eleven and rerun the cell, then restore ten and rerun before continuing.

```python
ordered = 12
reported_received = 10
difference = ordered - reported_received
print(difference)
```

## 2. Keep the source with the count

A dictionary holds named entries inside curly brackets. Follow the two keys in the first line. `"quantity"` holds a number. `"source_id"` holds text that identifies the record. Predict both lines of output. In the following cell, you will see what `.get("quantity")` returns when the key is absent.

```python
customer_report = {"quantity": 10, "source_id": "customer-17"}
reported_received = customer_report.get("quantity")
print(reported_received)
print(customer_report["source_id"])
```

## 3. Handle a missing count

This report has a source identifier but no quantity. `.get("quantity")` returns `None`, a special value for missing information. The `if` line checks for it. When the check succeeds, Python runs the indented line immediately below it. The `else` line introduces what happens when the condition does not hold. Predict which branch runs.

```python
incomplete_report = {"source_id": "customer-17"}
reported_received = incomplete_report.get("quantity")

if reported_received is None:
    print("Customer count missing. Ask for the quantity.")
else:
    print(ordered - reported_received)
```

## 4. Reuse the comparison

The function gives the comparison a name. Read the two lines at the bottom first. The first call passes twelve and ten into the function. The second passes twelve and `None`. Find the `return` line that each call reaches. `return` hands a result back to the caller. Here, the caller uses `print` to show it.

```python
def compare_quantities(ordered_boxes, reported_boxes):
    if reported_boxes is None:
        return "Customer count missing. Ask for the quantity."
    return ordered_boxes - reported_boxes

print(compare_quantities(12, 10))
print(compare_quantities(12, None))
```
