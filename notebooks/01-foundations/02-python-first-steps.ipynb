## Names, values and containers

`=` assigns a value to a name. An integer is a whole number. A string is text in quotation marks. A list holds an ordered sequence; a dictionary associates names with values. `print` displays a value. These structures let a program keep quantities and sources separate.


```python
ordered = 12
reported = 10
difference = ordered - reported
record = {"quantity": reported, "source_id": "customer-17"}
sources = ["order-17", record["source_id"]]
print(difference, sources)
assert difference == 2
```

## Functions and return values

`def` defines a reusable operation. Indentation identifies its body. `return` sends a result back to the caller. This helper illustrates arithmetic only; unlike the full lab function, it does not validate inputs or preserve evidence status.


```python
def difference_between(order_quantity, reported_quantity):
    return order_quantity - reported_quantity

print(difference_between(12, 10))
assert difference_between(8, 9) == -1
```

## Loops and decisions

`for` repeats an operation for each item. `if` chooses a branch. A comparison such as `!=` checks whether two values differ. The printed warning is a prompt for investigation, not a conclusion about who is correct.


```python
for name in ["order", "shipment", "customer_report"]:
    print(name, lab.CASE[name]["quantity"])
if lab.CASE["order"]["quantity"] != lab.CASE["customer_report"]["quantity"]:
    print("Records disagree: investigate.")
```
