```python
import copy

changed_case = copy.deepcopy(lab.CASE)
changed_case["customer_report"]["quantity"] = 13
changed_brief = lab.build_brief(changed_case)
assert changed_brief["comparison"]["order_minus_customer_report"] == -1
assert changed_brief["permitted_output"] == "investigation_brief_only"
print(changed_brief["comparison"])

duplicated = copy.deepcopy(lab.CASE)
duplicated["shipment"]["source_id"] = duplicated["order"]["source_id"]
try:
    lab.build_brief(duplicated)
except ValueError as error:
    print("Expected rejection:", error)
else:
    raise AssertionError("Duplicated source identifiers should be rejected.")
```

## Worked interpretation and completion check
