## Compare work rather than celebrating a successful run

These invented observations illustrate a comparison plan. They are not research findings or a measured deployment. Each entry is a number of minutes per case. The groups are tiny and are not a causal experiment. Lower handling time alone cannot show that investigation quality or service outcomes improved.


```python
from statistics import mean

manual_minutes = [18, 24, 21, 17]
assisted_minutes = [14, 17, 16, 15]
print("Illustrative manual mean:", mean(manual_minutes))
print("Illustrative assisted mean:", mean(assisted_minutes))
print("Illustrative difference:", mean(manual_minutes) - mean(assisted_minutes))
assert mean(manual_minutes) - mean(assisted_minutes) == 4.5
```

## Worked answer

A comparison could record unsupported closure decisions and reopened cases alongside total elapsed resolution time.
