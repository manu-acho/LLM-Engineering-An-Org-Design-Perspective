Which output establishes that the customer physically received ten boxes? Before opening the answer below, identify what the program knows, what it calculated and what remains unknown.

## Worked answer

None establishes the actual received quantity. The customer report says ten, the order and shipment records say twelve, and subtraction yields two. Generating text from unrelated training sentences cannot resolve this disagreement. A service needs a way to investigate evidence, assign responsibility and authorise any remedy. This distinction is the organisational starting point, rather than a detail to add after choosing a model.
