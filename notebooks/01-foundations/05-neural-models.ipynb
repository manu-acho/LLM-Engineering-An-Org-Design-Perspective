## Convert supplied scores into probabilities

A neural language model produces scores over its vocabulary. This notebook does not load or train a neural network. It isolates one calculation, softmax, using three supplied numbers. Subtracting the largest score before exponentiation helps avoid numerical overflow. Softmax probabilities concern continuation choices; they are not probabilities that a delivery claim is true.


```python
scores = [1.0, 0.0, -1.0]
values = lab.softmax(scores)
print(values)
assert abs(sum(values) - 1.0) < 1e-12
```

## Temperature changes relative concentration

A lower positive temperature makes score differences more influential. A higher temperature makes the distribution flatter. The function rejects zero; a separate greedy choice is used when the intended operation is simply to select the maximum.


```python
for temperature in [0.5, 1.0, 2.0]:
    print(temperature, lab.softmax(scores, temperature=temperature))
assert lab.softmax(scores, 0.5)[0] > lab.softmax(scores, 2.0)[0]
```

## Numerical stability

Adding the same constant to every score should leave the distribution unchanged. Floating-point calculations approximate real numbers, so comparisons use a tolerance. This checks a mathematical property, rather than trusting a plausible-looking printout.


```python
import math
shifted = lab.softmax([1001.0, 1000.0, 999.0])
assert all(math.isclose(a, b, abs_tol=1e-12) for a, b in zip(values, shifted))
print(shifted)
```

## Exercise and worked answer

Would a lower temperature verify that the order record is correct? No. Temperature changes a distribution over generated continuations. It does not create new evidence, alter decision rights or validate the sources.
