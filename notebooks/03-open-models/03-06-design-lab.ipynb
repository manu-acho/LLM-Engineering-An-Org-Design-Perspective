## Worked answer

Pretrained weights contain relationships estimated from prior training. Random weights do not. Both produce scores through a forward pass. A pretrained continuation still requires task evaluation and cannot establish who has authority to approve a meeting record. Hosting a model locally also does not confer a licence to redistribute its weights.
