## Worked answer

The function cannot judge entailment. A deterministic quantity comparison, a reviewer or a separately evaluated evidence-support check must examine whether the source supports the conclusion. Each additional checker has its own error modes. An acceptance contract should state exactly which checks were applied and what remains for review.

```python
from ai_orgdesign.labs.m10 import accept
try:
    accept({"case_id":"H-17", "sources":["invented"], "conclusion":"ok", "unresolved":["receipt"]})
except ValueError:
    print("Unsupported source rejected")
else:
    raise AssertionError("Acceptance failed")
```

## Alternatives and organisational meaning

A specialist-as-tool returns a contribution to its coordinator. A handoff transfers the active interaction to another role. Neither pattern should leave ownership implicit. Plain functions are adequate for many fixed routes and make a useful comparison before introducing a model-driven coordinator.
