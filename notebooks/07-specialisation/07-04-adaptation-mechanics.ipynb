## Worked programming solution

Compare the operations rather than copying the result. Explain why each contrasting example should produce its expected value.


```python
def masked_mean(values, mask):
    if not sum(mask):
        raise ValueError("No selected positions")
    return sum(value*selected for value,selected in zip(values,mask))/sum(mask)

assert masked_mean([10.,2.],[0,1]) == 2.
assert masked_mean([10.,2.],[1,1]) == 6.
```
