## Worked answer

Selection decisions convey information about the test set. Repeated optimisation against its scores can overfit the selection process without directly displaying labels. Keep separate development and final confirmation sets, restrict evaluation access, record the number of selection rounds and use a new representative comparison when scope changes. Human approval should examine the experiment design as well as the displayed score.

```python
from ai_orgdesign.labs.m20 import gate
baseline = {"n":10,"dataset_hash":"a","accuracy":.8,"unsafe":1,"cost":1.0}
assert not gate(baseline, {**baseline,"accuracy":.9,"unsafe":2}, True)["promote"]
print("Safety regression blocks promotion")
```

## Alternatives and organisational meaning

A fixed, periodically reviewed configuration may be safer and cheaper than automatic proposal generation. An improver can suggest changes without holding promotion authority. Versioning and rollback preserve a recovery route, but cannot undo every external consequence already caused by a released version.
