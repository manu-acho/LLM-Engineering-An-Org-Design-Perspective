## Read the new syntax and concepts first

A delta is the new part of an output. A generator yields successive pieces while retaining its place. A completed stream includes a final response status; an interrupted one leaves partial text.


## Execute the written lesson's mechanism

The following code is the worked mechanism printed in the notes. Run it before the practical extension. It is separate from any live request and makes the written explanation executable.

```python
def accumulate(deltas):
    text = ""
    for delta in deltas:
        text += delta
        yield text

assert list(accumulate(["Order", " O-17"])) == ["Order", "Order O-17"]
```
